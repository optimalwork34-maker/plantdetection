```python
#PRACTICAL 3 Python Programming for Cleaning the Data.
import pandas as pd

# 1. Load the dataset
df = pd.read_csv("heart.csv")

# 2. Display first 5 rows
print(df.head())

# 3. Check the size of the dataset
print("Dataset Shape:", df.shape)

# 4. Clean column names
df.columns = df.columns.str.strip().str.lower()

# 5. Check missing values
print("Missing Values:")
print(df.isnull().sum())

# 6. Check duplicate rows
print("Duplicate Rows:", df.duplicated().sum())

# 7. Remove duplicate rows
df = df.drop_duplicates()

# 8. Fill missing numerical values with mean
for column in df.select_dtypes(include="number"):
    df[column] = df[column].fillna(df[column].mean())

# 9. Fill missing text values with mode
for column in df.select_dtypes(exclude="number"):
    df[column] = df[column].fillna(df[column].mode()[0])

# 10. Check the cleaned dataset
print("Final Missing Values:")
print(df.isnull().sum())

print("Final Duplicate Rows:")
print(df.duplicated().sum())

# 11. Display cleaned data
print(df.head(10))

# 12. Save the cleaned dataset
df.to_csv("cleaned_heart.csv", index=False)

print("Data cleaning completed successfully!")
```
