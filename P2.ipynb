```python
#PRACTICAL 2 Write a Python program to load a dataset from a CSV file and display the data, first five records, column names, and basic information about the dataset.

import pandas as pd

# Load the dataset
data = pd.read_csv("heart.csv")

# Display complete dataset
print("Complete Dataset:")
print(data)

# Display first five records
print("\nFirst Five Records:")
print(data.head())

# Display column names
print("\nColumn Names:")
print(data.columns)

# Display information about dataset
print("\nDataset Information:")
print(data.info())
```
